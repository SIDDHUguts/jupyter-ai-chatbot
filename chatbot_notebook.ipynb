## 📊 Step 3: Chat with your Data (DataFrame Context Injection)

Let's create a sample pandas DataFrame and attach it to the chatbot context.

```python
# Create a sample dataset for analysis
import pandas as pd
import numpy as np

np.random.seed(42)
dates = pd.date_range('2026-01-01', periods=12, freq='M')
df_sales = pd.DataFrame({
    'Month': dates.strftime('%B'),
    'Revenue_USD': np.random.randint(45000, 120000, size=12),
    'Units_Sold': np.random.randint(300, 1200, size=12),
    'Satisfaction_Score': np.round(np.random.uniform(4.0, 5.0, size=12), 2)
})

print("📊 Generated Sample Sales Dataset:")
df_sales
```

### 📎 Attach `df_sales` to the Engine and Query

```python
# Initialize an engine with Data Science persona
ds_engine = ChatbotEngine(default_persona="Data Science & ML Pro")

# Attach the dataframe
ds_engine.attach_variable("df_sales", df_sales)

# Ask the chatbot to analyze it
analysis_reply = ds_engine.send_message("Analyze the attached df_sales and suggest visualization code.")
print(analysis_reply)
```

## 🌊 Step 4: Streaming Generation (Live Output)
You can stream tokens in real-time as the assistant generates text:

```python
# Real-time token streaming demo
print("Streaming Response:")
for token in ds_engine.send_message_stream("What are 3 key metrics to calculate for sales performance?"):
    sys.stdout.write(token)
    sys.stdout.flush()
```

## 💾 Step 5: Save & Restore Conversation Sessions
Save conversation threads to disk and restore them later:

```python
# Save session to JSON
session_file = "sessions/sales_analysis_demo.json"
ds_engine.save_session(session_file, session_name="Sales Analysis Demo")
print(f"💾 Saved session to {session_file}")

# Load session into a fresh engine
restored_engine = ChatbotEngine()
restored_engine.load_session(session_file)
print(f"📂 Restored session with {len(restored_engine.history)} messages!")
```
